## Extend Bucket Sort to $[a,b]$

Complete the following program by replacing the **10 blanks**. The blanks are written as `____1____`, `____2____`, and so on.


```python
def bucket_sort_ab(A, a, b):
    n = len(A)

    # Width of each bucket
    width = (b - a) / n

    # Create n empty buckets
    buckets = [[] for i in range(n)]

    # Place each element in a suitable bucket
    for x in A:
        index = min(
            int(n * (x - a)
                / (b - a)),
            n - 1
        )
        buckets[index].append(x)

    # Display the interval and contents of each bucket
    for i in range(n):
        left = a + i * width
        right = left + width
        print("Bucket", i, "[", left, ",", right, "]:", buckets[i])

    # Sort each bucket
    for bucket in buckets:
        bucket.sort()

    # Combine the sorted buckets
    result = []
    for bucket in buckets:
        result.extend(bucket)

    return result
```

```python
def insertion_sort(bucket):
    for i in range(1, len(bucket)):
        key = bucket[i]
        j = i - 1

        while j >= 0 and bucket[j] > key:
            bucket[j + 1] = bucket[j]
            j -= 1

        bucket[j + 1] = key


def bucket_sort_ab(A, a, b):
    n = len(A)

    if n == 0:
        return []

    width = (b - a) / n
    buckets = [[] for i in range(n)]

    # Distribute elements
    for x in A:
        index = min(
            int(n * (x - a) / (b - a)),
            n - 1
        )
        buckets[index].append(x)

    print("Buckets before sorting:")
    for i in range(n):
        print("Bucket", i, ":", buckets[i])
        
    # Sorting using insertion sort
    for bucket in buckets:
        insertion_sort(bucket)

    print("\nBuckets after sorting:")
    for i in range(n):
        print("Bucket", i, ":", buckets[i])

    result = []
    for bucket in buckets:
        result.extend(bucket)

    return result

d = 8
a = 10 + d
b = 20 + d

A = [19 + d, 12 + d, 16 + d, 11 + d,
     18 + d, 14 + d, 20 + d, 15 + d]

print("Registration Number: 240957062")
print("Digit Root:", d)
print("Original list:", A)
print("Interval: [", a, ",", b, "]")

result = bucket_sort_ab(A, a, b)

print("\nFinal sorted list:", result)
```

```python
def insertion_sort(bucket):
    for i in range(1, len(bucket)):
        key = bucket[i]
        j = i - 1
        while j >= 0 and bucket[j][0] > key[0]:
            bucket[j + 1] = bucket[j]
            j -= 1

        bucket[j + 1] = key


def bucket_sort_ab(A_tuples, a, b):
    n = len(A_tuples)

    if n == 0:
        return []

    width = (b - a) / n
    buckets = [[] for i in range(n)]

    # Distribute elements
    for item in A_tuples:
        x = item[0]
        index = min(
            int(n * (x - a) / (b - a)),
            n - 1
        )
        buckets[index].append(item)

    print("Buckets before sorting:")
    for i in range(n):
        print("Bucket", i, ":", buckets[i])
        
    # Sorting using insertion sort
    for bucket in buckets:
        insertion_sort(bucket)

    print("\nBuckets after sorting:")
    for i in range(n):
        print("Bucket", i, ":", buckets[i])

    result = []
    for bucket in buckets:
        result.extend(bucket)

    return result


d = 8
a = 10 + d
b = 20 + d

A = [12 + d, 12 + d, 16 + d, 11 + d,
     18 + d, 14 + d, 20 + d, 15 + d]

print("Registration Number: 240957062")
print("Digit Root:", d)
print("Original list:", A)
print("Interval: [", a, ",", b, "]")

# Creating a separate tuple structure
A_tuples = [(val, idx) for idx, val in enumerate(A)]
print("Original tuple list:", A_tuples)

result = bucket_sort_ab(A_tuples, a, b)

print("\nFinal sorted list:", result)
```
