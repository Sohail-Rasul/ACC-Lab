````python
def maximum_sum_recursive(triangle, row=0, col=0):
    if row == len(triangle) - 1:
        return triangle[row][col]
    left = maximum_sum_recursive(triangle, row + 1, col)
    right = maximum_sum_recursive(triangle, row + 1, col + 1)
    return triangle[row][col] + max(left, right)
print("Brute-force result for 4 rows:", maximum_sum_recursive(small_triangle))
````

### Part D: Dynamic Programming

````python
def maximum_sum_dp(triangle):
    dp = [row[:] for row in triangle]
    for i in range(len(triangle) - 2, -1, -1):
        for j in range(len(triangle[i])):
            dp[i][j] = triangle[i][j] + max(dp[i + 1][j],
                                             dp[i + 1][j + 1])
    return dp[0][0], dp
maximum_sum, dp = maximum_sum_dp(triangle)
print("Maximum sum for 44 rows:", maximum_sum)
def find_maximum_path(triangle):
    dp = [row[:] for row in triangle]
    for i in range(len(triangle) - 2, -1, -1):
        for j in range(len(triangle[i])):
            dp[i][j] = triangle[i][j] + max(dp[i + 1][j],
                                             dp[i + 1][j + 1])
    path = [triangle[0][0]]
    col = 0
    for i in range(len(triangle) - 1):
        if dp[i + 1][col] >= dp[i + 1][col + 1]:
            col = col
        else:
            col = col + 1
        path.append(triangle[i + 1][col])
    return path
maximum_path = find_maximum_path(triangle)
print("One maximum-sum path:")
print(maximum_path)
print("Check sum:", sum(maximum_path))
````

### Part E: Pascal's Triangle

````python
def make_pascal(n):
    triangle = []
    for i in range(n):
        row = []
        for j in range(i + 1):
            row.append(math.comb(i, j))
        triangle.append(row)
    return triangle
def pascal_max_path(n):
    triangle = make_pascal(n)
    dp = [row[:] for row in triangle]
    for i in range(n - 2, -1, -1):
        for j in range(i + 1):
            dp[i][j] = triangle[i][j] + max(dp[i + 1][j],
                                             dp[i + 1][j + 1])
    path = [triangle[0][0]]
    col = 0
    for i in range(n - 1):
        if dp[i + 1][col] >= dp[i + 1][col + 1]:
            col = col
        else:
            col = col + 1
        path.append(triangle[i + 1][col])
    return path, dp[0][0]
for n in [5, 10, 15, 20]:
    path, total = pascal_max_path(n)
    print("\nn =", n)
    print("Maximum-sum path:", path)
    print("Maximum sum:", total)
````

##### Part E:

For Pascal's triangle, the maximum sums are 13, 274, 7060, 191673 for n=5,10,15,20, respectively. 

Observations / conjectures:

 1. A maximum path stays close to one side of Pascal's triangle,
    moving toward larger binomial coefficients.

 2. The maximum path is NOT unique. Pascal's triangle is symmetric,
    so reflecting a maximum path gives another maximum path.
    Equal choices inside the triangle can create still more paths.

 3. The maximum sums for n = 5, 10, 15, 20 are:
       13, 274, 7060, 191673

    There is no simple arithmetic/geometric pattern visible from
    these four values. A useful general conjecture is that the
    maximum sum is obtained by the dynamic-programming recurrence
       M(i,j) = C(i,j) + max(M(i+1,j), M(i+1,j+1))
    with the maximum value at the top entry.

    In other words, the DP recurrence gives a reliable way to find
    the maximum for any n, even when a simple closed formula is not
    apparent.
