## Question 1
Study the program. What is the program doing? How many passes (iterations of i loop) are executed?

Note: range(k) goes from 0 to k-1.

Answer:
Bubble sorting in ascending order. It compares adjacent elements and swaps them.

Range(3) => 3 passes


```python
a=[5,1,4,2]
for i in range(3):
    for j in range(3-i):
        if a[j]>a[j+1]:
            a[j],a[j+1]=a[j+1],a[j]
print(a)
```

## Question 10
Determine what the variable count stores and find its final value.

Answer:

count stores the total number of swaps

Pass 1: 3 swaps

4,3,2,1 -> 3,4,2,1 -> 3,2,4,1 -> 3,2,1,4

Pass 2: 2 swaps

Total swaps: 3 + 2 + 1 = 6

Output: 6

## Question 11

Write the code for optimized bubble sort using the idea in Question 7. Illustrate with an example.

```python
a = [5,1,4,2]
n = len(a)

for i in range(n-1):
    flag = 0
    for j in range(n-1-i):
        if a[j] > a[j+1]:
            a[j], a[j+1] = a[j+1], a[j]
            flag = 1
    if flag == 0:
        break

print(a)
```
